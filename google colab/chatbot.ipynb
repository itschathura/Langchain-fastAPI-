```python
#import necessary libraries
import os
from google.colab import userdata
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")

model = ChatGoogleGenerativeAI(
    model="gemini-3.7-flash",
    temperature=1.0,  # Gemini 3.0+ defaults to 1.0
    max_tokens=None,
    timeout=None,
    max_retries=2,
    # other params...
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

#create a prompt template
prompt = ChatPromptTemplate(
    [
        ("system", "You are a helpful AI bot. Your name is {name}."),
        ("human", "Hello, how are you doing?"),
        ("ai", "I'm doing well, thanks!"),
        ("human", "{user_input}"),
    ]
)
```

```python
#chain the prompt , llm and parser
# chain = prompt | model     *********** without parser

from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

chain = prompt | model | StrOutputParser()
```

```python
user_input = "My name is Bola"

response = chain.invoke({
    "name": "Bob",
    "user_input": user_input
})

print(response)
```

# Initialize Prompt template for Dynamic Interaction

```python
from langchain_core.prompts.chat import MessagesPlaceholder
from langchain_core.prompts import MessagesPlaceholder
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

#create a prompt template using MessagesPlaceholder for the question

chat_history = [
    HumanMessage(content="My name is Bola."),
    AIMessage(content="Nice to meet you, Bola!")
]
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate([
    ("system", "You are an intelligent chatbot. Answer the following questions."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{question}")
])
```

```python
chain = prompt | model | StrOutputParser()
```

```python

response2 = chain.invoke({
    "chat_history": chat_history,
    "question": "What is my name?"
})

print(response2)
```

```python
test_prompt = prompt.invoke({
    "chat_history": chat_history,
    "question": "What is my name?"
})

print(test_prompt.messages)
```
